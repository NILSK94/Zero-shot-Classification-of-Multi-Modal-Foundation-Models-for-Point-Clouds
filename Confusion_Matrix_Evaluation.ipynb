# Evaluation

Set the CSV and figure paths. The CSV needs image and label columns.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_recall_fscore_support,
    classification_report
)

# Load CSV file
csv_path = r"XXXXXXXXXXXXX"  # INPUT PATH FROM THE Classified Data of the CSV
df = pd.read_csv(csv_path)

# Extract true and predicted labels
df["true"] = df["image"].str.split("_").str[0].str.lower()  # HERE YOU NEED TO ADD THE First row of the CSV or change it in the CSV to "image,label"
df["pred"] = df["label"].str.strip().str.lower()            # HERE YOU NEED TO ADD THE First row of the CSV or change it in the CSV to "image,label"

# Determine all unique class labels
classes = sorted(set(df["true"]) | set(df["pred"]))

# Compute confusion matrix
cm = confusion_matrix(df["true"], df["pred"], labels=classes)

# Display confusion matrix
fig, ax = plt.subplots(figsize=(10, 10))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=classes)
disp.plot(include_values=True, cmap="Blues", ax=ax, xticks_rotation=90)
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()

# Save confusion matrix as PNG
output_path = r"XXXXXXXXX"   # SAFE PATH OF THE CONFUSION MATRIX
fig.savefig(output_path, dpi=300)

# Compute overall accuracy
accuracy = accuracy_score(df["true"], df["pred"])
print(f"\nOverall Accuracy: {accuracy:.2%}")

# Compute per-class metrics
precision, recall, f1, support = precision_recall_fscore_support(
    df["true"], df["pred"], labels=classes, zero_division=0
)

# Print per-class evaluation
print("\nPer-Class Evaluation:")
for cls, p, r, f, s in zip(classes, precision, recall, f1, support):
    print(f"- {cls:<12} | Precision: {p:.2%} | Recall: {r:.2%} | F1 Score: {f:.2%} | Support: {s}")

# Print full classification report
print("\nClassification Report:")
print(classification_report(df["true"], df["pred"], labels=classes, zero_division=0))
```
