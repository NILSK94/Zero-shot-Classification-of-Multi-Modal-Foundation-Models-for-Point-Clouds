# ModelNet classification

Set input paths and GEMINI_API_KEY. Run the cell for 1, 2, 3, or 4 views.

```python
import os
import time
import csv
import shutil
from pathlib import Path
from PIL import Image
import google.generativeai as genai

# API 
GOOGLE_API_KEY = os.getenv("GEMINI_API_KEY", "")
if not GOOGLE_API_KEY:
    raise ValueError("Missing GOOGLE_API_KEY. Please configure it before execution.")

genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel("gemini-1.5-pro")

# Directory 
input_dir = Path(r"XXXXX")  # YOUR INPUT FOLDER
output_dir = Path(r"XXXXX") # YOUR OUTPUT FOLDER
csv_file = output_dir / "classification_gemini_single_image.csv"
delay_between_requests = 1.5  # seconds

# Class list
class_list = {
    "airplane", "bed", "bench", "bookshelf", "bottle", "bowl", "car", "chair",
    "cone", "cup", "curtain", "door", "flower_pot", "glass_box", "guitar",
    "keyboard", "lamp", "laptop", "mantel", "person", "piano", "plant", "radio",
    "range_hood", "sink", "stairs", "stool", "tent", "toilet", "tv_stand",
    "vase", "wardrobe", "xbox"
}

# Prompt for classification
prompt = (
    "Classify the object shown in one isometric views. "
    "Select exactly one category from the list below that best describes the object. "
    "Respond with only the category name (no additional text):\n\n"
    f"{', '.join(sorted(class_list))}"
)

def find_label_in_response(text: str) -> str | None:
    text = text.strip().lower()
    for label in class_list:
        if f" {label} " in f" {text} " or text == label:
            return label
    return None

def classify_image(image_path: Path) -> str:
    image = Image.open(image_path)
    response = model.generate_content([prompt, image])
    label = find_label_in_response(response.text)
    if label is None:
        raise ValueError(f"No valid label found in model response: '{response.text}'")
    return label

def save_result(image_name: str, label: str):
    write_header = not csv_file.exists()
    with open(csv_file, mode="a", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        if write_header:
            writer.writerow(["image", "label"])
        writer.writerow([image_name, label])

def main():
    image_paths = list(input_dir.rglob("*.png"))
    total = len(image_paths)

    if total == 0:
        print(f"No PNG images found in directory: {input_dir}")
        return

    print(f"{total} images found. Starting classification...")

    for index, image_path in enumerate(image_paths):
        image_name = image_path.name
        print(f"[{index + 1}/{total}] Processing: {image_name}")

        try:
            label = classify_image(image_path)
            print(f"Classified as: {label}")

            label_dir = output_dir / label
            label_dir.mkdir(parents=True, exist_ok=True)
            shutil.copy(image_path, label_dir / image_name)

            save_result(image_name, label)

        except Exception as e:
            print(f"Error processing image '{image_name}': {e}")

        if index + 1 < total:
            time.sleep(delay_between_requests)

    print("Classification completed.")

if __name__ == "__main__":
    main()
```
